```python

import pandas as pd
import ollama

df = pd.read_csv("../data/prompts.csv")

responses = []

for prompt in df["prompt"]:
    result = ollama.generate(
        model="llama3.2",
        prompt=prompt
    )

    responses.append({
        "prompt": prompt,
        "response": result["response"]
    })

    print("Prompt:", prompt)
    print("Response:", result["response"])
    print("-" * 40)

output = pd.DataFrame(responses)
output.to_csv("../data/notebook_responses.csv", index=False)

print("All notebook responses saved successfully!")
```
